```python
# Nama : Sri Puspita
# Nim : 4232501015
# Lima latihan kecil (data kecil)

skor = [80, 65, 90, 75, 70, 85]

print("Skor pertama:", skor[0])
print("Skor terakhir:", skor[-1])

rata_rata = sum(skor) / len(skor)
print("Rata-rata:", rata_rata)

jumlah = 0
for nilai in skor:
    if nilai > rata_rata:
        jumlah += 1

print("Skor di atas rata-rata:", jumlah)

terendah = min(skor)
posisi = skor.index(terendah) + 1
print("Nomor mahasiswa dengan skor terendah:", posisi)

skor_baru = [nilai + 5 for nilai in skor]
print("Skor baru:", skor_baru)
```

```python
# Nama : Sri Puspita
# Nim : 4232501015
# Laporan produksi bulanan (portofolio)

energi = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062
]

target = float(input("Masukkan target energi: "))

total = sum(energi)
rata_rata = total / len(energi)

print("\nTotal energi:", total)
print("Rata-rata energi:", rata_rata)

data = [(nilai, i+1) for i, nilai in enumerate(energi)]
tertinggi = sorted(data, reverse=True)[:3]

print("\nTiga hari tertinggi:")
for nilai, hari in tertinggi:
    print("Hari", hari, ":", nilai)

terendah = sorted(data)[:3]

print("\nTiga hari terendah:")
for nilai, hari in terendah:
    print("Hari", hari, ":", nilai)

jumlah = sum(1 for nilai in energi if nilai < target)

print("\nJumlah hari di bawah target:", jumlah)

batas = rata_rata * 0.97

print("\nRata-rata bergerak di bawah 97% rata-rata bulanan:")

for i in range(len(energi) - 6):
    rata7 = sum(energi[i:i+7]) / 7

    if rata7 < batas:
        print("Hari", i+1, "-", i+7, ":", round(rata7, 2))

print("\nTotal energi per minggu:")

for i in range(0, len(energi), 7):
    minggu = energi[i:i+7]
    print("Hari", i+1, "-", i+len(minggu), ":", sum(minggu))
```
