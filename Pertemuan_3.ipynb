

1. Urutan Prioritas Operator di Python
# Operator precedence adalah aturan yang digunakan Python untuk menentukan operator mana yang dikerjakan terlebih dahulu ketika terdapat beberapa operator dalam satu ekspresi. Berdasarkan dokumentasi resmi Python, setiap operator mempunyai tingkat prioritas yang berbeda. Operator yang mempunyai prioritas lebih tinggi akan dikerjakan terlebih dahulu.

# Beberapa urutan operator dari prioritas tinggi ke rendah
# () Tanda kurung
# ** Perpangkatan
# *, /, //, % Perkalian dan pembagian
# +, - Penjumlahan dan pengurangan
# >, <, >=, <=, ==, != Perbandingan
# not Logika NOT
# and Logoika AND
# or Logika OR


# Mengapa ekspresi 10 + 5 * 2 menghasilkan 20, bukan 30?
Menurut saya, karena operator precedence itu penting karena Python memiliki aturan dalam menentukan operasi mana yang harus dikerjakan terlebih dahulu. Jadi, Python tidak langsung menghitung dari kiri ke kanan, tetapi melihat operator yang memiliki prioritas lebih tinggi. Pada contoh 10 + 5 * 2, Python mengerjakan perkalian terlebih dahulu, yaitu 5 × 2 = 10. Setelah itu, hasilnya saya tambahkan dengan 10, jadi 10 + 10 = 20.

2. Perbedaan Operator Relasional vs Operator Logika

#Operator Relasional (>, <, ==, >=, <=, !=): Membandingkan dua nilai (operand) untuk menghasilkan nilai boolean (True atau False).
Contoh: nilai > 70.
#​Operator Logika (and, or, not): Menggabungkan atau memanipulasi satu atau lebih ekspresi boolean untuk menghasilkan keputusan logis.
Contoh: a and b.

3. Contoh Kasus Bug akibat Kesalahan Operator Precedence

#Kasus:
#Seorang pengembang perangkat lunak ingin menghitung rata-rata dari tiga variabel (a, b, dan c). Pengembang menuliskan kode rata_rata = a + b + c / 3. Karena operator pembagian (/) memiliki prioritas lebih tinggi daripada penjumlahan (+), Python hanya membagi c dengan 3, lalu menambahkan hasilnya ke a dan b. Hal ini menyebabkan kalkulasi rata-rata salah dan menghasilkan bug logika pada aplikasi keuangan.

Gunakan tanda kurung () untuk mengelompokkan penjumlahan terlebih dahulu sebelum pembagian:
# Kode yang benar:
rata_rata = (a + b + c) / 3
