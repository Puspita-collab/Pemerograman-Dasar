```python
# Nama : Sri Puspita
# Nim : 4232501015
# Program : Soal 1: Belanja Buku

jumlah_buku = int(input("Masukkan jumlah buku: "))
harga_satuan = int(input("Masukkan harga per buku: "))
persen_diskon = float(input("Masukkan persentase diskon: "))
uang_budi = int(input("Masukkan jumlah uang Budi: "))
jumlah_orang = int(input("Masukkan jumlah orang yang patungan: "))

total = jumlah_buku * harga_satuan
diskon = int(total * persen_diskon)
bayar = total - diskon
kembalian = uang_budi - bayar
per_orang = bayar // jumlah_orang
sisa = bayar % jumlah_orang
uang_cukup = uang_budi >= bayar

print("\n=== HASIL PERHITUNGAN Belanja Buku ===")
print(f"Total       : {total}")
print(f"Diskon      : {diskon}")
print(f"Bayar       : {bayar}")
print(f"Kembalian   : {kembalian}")
print(f"Per orang   : {per_orang}")
print(f"Sisa        : {sisa}")
print(f"Uang cukup? : {uang_cukup}")
```

```python
# Nama : Sri Puspita
# Nim : 4232501015
# Program : Soal 2: PLTS Atap Rumah

jumlah_panel = int(input("Masukkan jumlah panel surya: "))
daya_panel = float(input("Masukkan daya panel: "))
jam_penyinaran = float(input("Masukkan jam penyinaran: "))
efisiensi = float(input("Masukkan efisiensi sistem: "))
kebutuhan_listrik = float(input("Masukkan kebutuhan listrik rumah: "))
tarif_listrik = float(input("Masukkan tarif listrik: "))
hari_bulan = int(input("Masukkan jumlah hari dalam sebulan: "))

energi_harian = (jumlah_panel * daya_panel * jam_penyinaran * efisiensi) / 1000
energi_bulanan = energi_harian * hari_bulan
nilai_listrik = energi_bulanan * tarif_listrik
kekurangan = kebutuhan_listrik - energi_harian
mencukupi = energi_harian >= kebutuhan_listrik

print("\n=== HASIL PERHITUNGAN PLTS Dan ATAP ===")
print(f"Energi harian (kWh) : {round(energi_harian, 1)}")
print(f"Energi bulanan (kWh): {round(energi_bulanan, 1)}")
print(f"Nilai listrik (Rp)  : {round(nilai_listrik, 1)}")
print(f"Kekurangan (kWh)    : {round(kekurangan, 1)}")
print(f"Mencukupi?          : {mencukupi}")
```

```python
# Nama : Sri Puspita
# Nim : 4232501015
# Program : Soal 3: Generator Diesel dan Rugi Kabel

daya_generator = float(input("Masukkan daya generator (kW): "))
jam_operasi = float(input("Masukkan jam operasi per hari: "))
konsumsi_solar = float(input("Masukkan konsumsi solar per kWh: "))
harga_solar = float(input("Masukkan harga solar per liter: "))
kapasitas_tangki = float(input("Masukkan kapasitas tangki (liter): "))
arus = float(input("Masukkan arus pada kabel (A): "))
hambatan = float(input("Masukkan hambatan kabel (ohm): "))

energi = daya_generator * jam_operasi
solar_per_hari = energi * konsumsi_solar
biaya = solar_per_hari * harga_solar
tangki_cukup = kapasitas_tangki // solar_per_hari
sisa_solar = kapasitas_tangki % solar_per_hari
rugi_kabel = (arus ** 2) * hambatan
rugi_tinggi = (rugi_kabel > 150) and (arus >= 20)

print("\n=== HASIL PERHITUNGAN ===")
print(f"Energi (kWh)      : {energi}")
print(f"Solar (liter)     : {solar_per_hari}")
print(f"Biaya (Rp)        : {biaya}")
print(f"Tangki cukup      : {tangki_cukup} hari")
print(f"Sisa solar (L)    : {sisa_solar}")
print(f"Rugi kabel (W)    : {rugi_kabel}")
print(f"Rugi tinggi?      : {rugi_tinggi}")
```
