```python
#Nama: Sri Puspita
#Nim : 4232501015
#Kelas: Pagi 3A
#H/T: Senin 09 November 2026
#Matakuliah: Pemrograman Dasar
#Soal 7 Pemanas Mencapai Target


suhu_saat_ini = 60

while suhu_saat_ini < 85:
    suhu_saat_ini += 5
    print(f"Suhu saat ini: {suhu_saat_ini} C")

    if suhu_saat_ini == 80:
        print("WARNING: Mendekati target!")

print("Target 85 C tercapai.")
```

```python
#Nama: Sri Puspita
#Nim : 4232501015
#Kelas: Pagi 3A
#H/T: Senin 09 November 2026
#Matakuliah: Pemrograman Dasar
#Soal 9 Pembacaan Vibrasi dengan Sensor Rusak


data = ["1.2", "0.9", "ERR", "3.1", "1.8"]

jumlah_tidak_valid = 0

for i in range(len(data)):
    try:
        nilai = float(data[i])

    except ValueError:
        print(f"Unit {i+1}: data tidak valid")
        jumlah_tidak_valid += 1
        continue

    if nilai > 2.0:
        print(f"Warning: Unit {i+1} vibrasi tinggi!")

    if nilai > 3.0:
        print("KRITIS: pemeriksaan dihentikan")
        break

print(f"Jumlah data tidak valid: {jumlah_tidak_valid}")
```

Program yang saya buat untuk memeriksa data vibrasi satu per satu. Data yang tidak bisa diubah menjadi angka akan dianggap tidak valid dan dihitung. Jika nilai vibrasi lebih dari 2.0, muncul peringatan. Jika lebih dari 3.0, program menampilkan kondisi kritis dan menghentikan pemeriksaan menggunakan break.




```python

# PROGRAM PEMANTAUAN KONSUMSI LISTRIK HARIAN
# Tujuan: Menghitung konsumsi listrik dan mendeteksi
# apakah penggunaan listrik termasuk hemat, normal, atau boros.

print("======================================")
print(" PROGRAM PEMANTAUAN KONSUMSI LISTRIK ")
print("======================================")

# List untuk menyimpan data konsumsi listrik selama 7 hari
konsumsi_listrik = []

# Perulangan untuk memasukkan data listrik selama 7 hari
for hari in range(1, 8):
    while True:
        try:
            energi = float(input(f"Masukkan konsumsi listrik hari ke-{hari} (kWh): "))

            if energi < 0:
                print("Konsumsi tidak boleh negatif!")
            else:
                konsumsi_listrik.append(energi)
                break

        except ValueError:
            print("Input salah! Masukkan angka.")

# Menghitung total konsumsi listrik
total = sum(konsumsi_listrik)

# Menghitung rata-rata konsumsi listrik
rata_rata = total / len(konsumsi_listrik)

print("\n========== HASIL PEMANTAUAN ==========")

# Menampilkan konsumsi listrik setiap hari
for i in range(len(konsumsi_listrik)):
    print(f"Hari ke-{i+1}: {konsumsi_listrik[i]} kWh")

# Menampilkan total dan rata-rata
print("--------------------------------------")
print(f"Total konsumsi   : {total:.2f} kWh")
print(f"Rata-rata harian : {rata_rata:.2f} kWh")

# Percabangan untuk mendeteksi penggunaan listrik
if rata_rata <= 5:
    print("Status: HEMAT ENERGI")
    print("Penggunaan listrik sangat baik.")
elif rata_rata <= 10:
    print("Status: NORMAL")
    print("Penggunaan listrik masih dalam batas wajar.")
else:
    print("Status: BOROS ENERGI")
    print("Disarankan mengurangi penggunaan listrik.")

print("======================================")
print("Terima kasih telah menggunakan program!")
```
